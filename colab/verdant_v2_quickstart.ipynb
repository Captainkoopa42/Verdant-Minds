```python
import matplotlib.pyplot as plt
from pathlib import Path
figs = [
    Path('results/20260306T045715Z/plots/depth_vs_age.png'),
    Path('results/20260306T045715Z/plots/older_to_newer_share.png'),
    Path('results/20260306T045715Z/plots/lineage_growth.png'),
    Path('results/20260306T045715Z/plots/emergent_distribution.png'),
]
for fig in figs:
    if fig.exists():
        img = plt.imread(fig)
        plt.figure(figsize=(8, 4.5))
        plt.imshow(img)
        plt.title(fig.name)
        plt.axis('off')
        plt.show()
    else:
        print(f'Missing figure: {fig}')
```
