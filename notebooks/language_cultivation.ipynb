# Verdant Language Cultivation
Feed phonemes, grammar, lexicon, and sentences through Verdant.


```python
from pathlib import Path
import json
from verdant.system import VerdantSystem
from verdant.io import HumanTextInputAdapter, ConversationalOutputAdapter, QueryInterface, InteractionService
from verdant.io.see_and_say import run_session
CHECKPOINT=None
system=VerdantSystem()
if CHECKPOINT:
    system.load_state(CHECKPOINT)
inp=HumanTextInputAdapter('will','lang_cult')
out=ConversationalOutputAdapter()
qi=QueryInterface(system)
system.register_adapter(inp)
system.register_output_adapter(out)
svc=InteractionService(system, inp, out, qi, base_dir='interactions')
```

```python
phonemes=json.loads(Path('corpus/phonemes/english_phoneme_inventory.json').read_text())
for p in phonemes:
    svc.turn('will', json.dumps(p))
print('fed phonemes', len(phonemes))
```

```python
rules=json.loads(Path('corpus/grammar/english_grammar_rules.json').read_text())
for r in rules:
    svc.turn('will', json.dumps(r))
print('fed rules', len(rules))
```

```python
lex=json.loads(Path('corpus/lexicon/core_vocabulary_physical_world.json').read_text())
for i,w in enumerate(lex,1):
    svc.turn('will', json.dumps(w))
    if i%25==0: print('fed words',i)
```

```python
sentences=json.loads(Path('corpus/sentences/teaching_sentences_physical_world.json').read_text())
for s in sentences:
    svc.turn('will', json.dumps(s))
print('fed sentences', len(sentences))
```

```python
result=run_session('corpus/lists/teaching_list_physical_world.json', qi)
print(result['pass_rate'])
```

```python
after_nodes=set(system.memory_web.list_concepts())
after_edges=system.memory_web.graph.number_of_edges()
print('nodes_total:', len(after_nodes), 'edges_total:', after_edges)
print('basin_count', system.get_metrics().get('basin_count'))
```
